```python
# -*- coding: utf-8 -*-
# 【CP1-03 Reducer 本质】自定义归约 vs add_messages 消息专用归约
# 文件：CP1/03_reducer.ipynb
# 作用：本 Cell 演示 【CP1-03 Reducer 本质】自定义归约 vs add_messages 消息专用归约 的完整可运行示例
# 阅读顺序：状态定义 → 节点定义 → 图构建 → 编译执行 → 结果观察

def my_reducer(left : list[str],right : list[str]) -> list[str]:
    return left + right

# node1运行后的值
node1_logs = ["node_1 运行完毕"]
# node2运行后的值
node2_logs = ["node_2 运行完毕"]
# 合并node1_logs和node2_logs
merged_logs = my_reducer(node1_logs,node2_logs)
print(merged_logs)
```

```python
# -*- coding: utf-8 -*-
# 【CP1-03 Reducer 本质】自定义归约 vs add_messages 消息专用归约
# 文件：CP1/03_reducer.ipynb
# 作用：本 Cell 演示 【CP1-03 Reducer 本质】自定义归约 vs add_messages 消息专用归约 的完整可运行示例
# 阅读顺序：状态定义 → 节点定义 → 图构建 → 编译执行 → 结果观察

"""AIMessage"""
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langgraph.graph.message import add_messages  # 消息归约：按 id 去重与时序合并


left = [
    SystemMessage(content="你是一个专业的翻译",id='1'),
    HumanMessage(content="你好",id='2'),
    AIMessage(content="你好，我是AI助手",id='3'),
]
right = [
    HumanMessage(content="你好",id='4'),
    AIMessage(content="你好，我是AI助手",id='5'),
]
merged=add_messages(left,right)  # 消息归约：按 id 去重与时序合并
print(merged)

```
