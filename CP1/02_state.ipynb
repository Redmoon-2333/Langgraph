```python
# -*- coding: utf-8 -*-
# 【CP1-02 State 三形态】dataclass / Pydantic / TypedDict 对比；揭示双重累加陷阱
# 文件：CP1/02_state.ipynb
# 作用：本 Cell 演示 【CP1-02 State 三形态】dataclass / Pydantic / TypedDict 对比；揭示双重累加陷阱 的完整可运行示例
# 阅读顺序：状态定义 → 节点定义 → 图构建 → 编译执行 → 结果观察

""" DATACLASS """
from typing import TypedDict,Annotated
from operator import add
from dataclasses import dataclass
from langgraph.graph import StateGraph,START,END

@dataclass
class OverAllState:
    logs: Annotated[list[str], add]
    cur_id:str

def node_1(state:OverAllState) -> OverAllState:
    pre_id = state.cur_id
    return OverAllState(state.logs + ["node_1 运行完毕"], pre_id + ", node_1")

def node_2(state:OverAllState) -> OverAllState:
    pre_id = state.cur_id
    return OverAllState(state.logs + ["node_2 运行完毕"], pre_id + ", node_2")

builder = StateGraph(state_schema=OverAllState)  # 创建状态图构建器：绑定状态 Schema

builder.add_node(node_1)  # 注册节点到图中
builder.add_node(node_2)  # 注册节点到图中

builder.add_edge(START,"node_1")  # 起点扇出
builder.add_edge("node_1","node_2")
builder.add_edge("node_2",END)  # 汇入终点

graph=builder.compile()  # 编译图：蓝图→可执行对象

result = graph.invoke({"cur_id":"start"})  # 触发图执行：传入初始 State + config
print(result)
```
